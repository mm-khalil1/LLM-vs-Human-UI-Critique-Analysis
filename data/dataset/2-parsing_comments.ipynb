# Pasre Comments to Structured Dicts

## Automatically parse all comments

```python
import re
import numpy as np
import pandas as pd
from typing import Any, Dict, List, Optional, Tuple

# ============================================================
# 0) Regex patterns (tiered, but strict about fix markers)
# ============================================================

# ---- Comment ID (optional) ----
RE_COMMENT_ID = re.compile(r"^\s*((?:llm\s+)?comment\s*[:#]?\s*\d+)\b[\s:,\-–—]*", re.I)

# ---- Bounding box ----
RE_BBOX = re.compile(r"(?:\bbounding\s*box\b|\bboundingbox\b)\s*[:\-]?\s*\[([^\]]+)\]", re.I)

# ---- Expected standard markers (primary + fallback variants you measured) ----
EXP_MARKERS = [
    re.compile(r"\bthe\s+expected\s+standard\b", re.I),   # primary
    re.compile(r"\bthe\s+standard\s+is\b", re.I),         # fallback
    re.compile(r"\bhe\s+expected\s+standard\s+is\b", re.I),    # typo: missing 't' (no newline required)
    re.compile(r"\nhe\s+expected\s+standard\s+is\b", re.I),    # typo with newline (kept)
]

# ---- Observed issue markers (primary + fallbacks you measured) ----
OBS_MARKERS = [
    re.compile(r"\bin\s+the\s+current\s+design\b", re.I),  # primary (comma not required)
    re.compile(r"\bin\s+the\s+current\s+layout\b", re.I),
    re.compile(r"\bin\s+this\s+current\s+design\b", re.I),
    re.compile(r"\bin\s+the\s+current\s+,", re.I),         # "in the current ,"
]

# ---- Fix markers (STRICT: ONLY "to fix this" or "to fix it") ----
RE_FIX_THIS = re.compile(r"\bto\s+fix\s+this\b", re.I)
RE_FIX_IT   = re.compile(r"\bto\s+fix\s+it\b", re.I)


# ============================================================
# 1) Small utilities
# ============================================================

def _strip_edges(s: str) -> str:
    return (s or "").strip()

def _norm_spaces(s: str) -> str:
    return re.sub(r"\s+", " ", (s or "").replace("\u00a0", " ")).strip()

def _strip_leading_punct(s: str) -> str:
    return re.sub(r"^[\s,:\-–—]+", "", (s or "").strip())

def _strip_trailing_punct(s: str) -> str:
    return re.sub(r"[\s,:\-–—]+$", "", (s or "").strip())

def _cleanup_segment(s: str) -> str:
    s = _strip_edges(s)
    s = _strip_leading_punct(s)
    s = _strip_trailing_punct(s)
    return s.strip()

def _extract_comment_id(text: str) -> Tuple[Optional[str], str]:
    m = RE_COMMENT_ID.match(text or "")
    if not m:
        return None, _strip_edges(text)
    cid = _strip_edges(m.group(1))
    cid = re.sub(r"^llm\b", "LLM", cid, flags=re.I)
    cid = cid[0:1].upper() + cid[1:]
    rest = _strip_edges((text or "")[m.end():])
    return cid, rest

def _extract_bbox(text: str) -> Tuple[str, Optional[List[float]]]:
    m = RE_BBOX.search(text or "")
    if not m:
        return text, None
    nums = re.findall(r"[-+]?\d*\.?\d+(?:[eE][-+]?\d+)?", m.group(1))
    bbox = [float(x) for x in nums] if nums else None
    text_wo = (text[:m.start()] + " " + text[m.end():]).strip()
    return _strip_edges(text_wo), bbox

def _find_earliest(text: str, patterns: List[re.Pattern]) -> Optional[re.Match]:
    best = None
    for p in patterns:
        m = p.search(text)
        if not m:
            continue
        if best is None or m.start() < best.start():
            best = m
    return best

def _remove_expected_glue(seg: str) -> str:
    seg = (seg or "")
    seg = seg.lstrip(" \t\r\n:,-–—")

    # remove in this exact priority:
    # 1) "is that"
    # 2) "is to"
    # 3) "that"
    # 4) "to"
    # 5) "is"
    seg = re.sub(r"^(is\s+that)\b\s*", "", seg, flags=re.I)
    seg = re.sub(r"^(is\s+to)\b\s*", "", seg, flags=re.I)
    seg = re.sub(r"^(that)\b\s*", "", seg, flags=re.I)
    seg = re.sub(r"^(to)\b\s*", "", seg, flags=re.I)
    seg = re.sub(r"^(is)\b\s*", "", seg, flags=re.I)

    return _cleanup_segment(seg) or ""


def infer_comment_type(d: Dict[str, Any]) -> str:
    has_exp = d.get("expected_standard") is not None
    has_obs = d.get("observed_issue") is not None
    has_fix = d.get("suggested_fix") is not None

    if has_exp and has_obs and has_fix:
        return "full"
    if has_exp and has_obs:
        return "expected_and_observed"
    if has_obs and has_fix:
        return "observed_and_fix"
    if has_exp and has_fix:
        return "expected_and_fix"
    if has_obs:
        return "observed_only"
    if has_fix:
        return "fix_only"
    if has_exp:
        return "expected_only"
    return "unclassified"


# ============================================================
# 2) Strict fix extraction (your exact requirement)
# ============================================================

def extract_fix_payload(text: str) -> Optional[str]:
    """
    STRICT:
      - Only reacts to "to fix this" or "to fix it"
      - Keeps only what comes AFTER the marker
      - If next word is "issue"/"issues", remove it
      - Then remove leading comma if present
    """
    if not text:
        return None

    m1 = RE_FIX_THIS.search(text)
    m2 = RE_FIX_IT.search(text)
    matches = [m for m in (m1, m2) if m]
    if not matches:
        return None

    m = min(matches, key=lambda x: x.start())
    rest = text[m.end():]  # after marker
    rest = rest.lstrip()

    # remove "issue"/"issues" if immediately next
    rest = re.sub(r"^(issue|issues)\b", "", rest, flags=re.I).lstrip()

    # remove leading comma if immediately next
    if rest.startswith(","):
        rest = rest[1:].lstrip()

    rest = _cleanup_segment(rest)
    return rest or None


# ============================================================
# 3) Core parsing for a single comment
# ============================================================

def parse_single_comment(comment: Any) -> Dict[str, Any]:
    raw = "" if comment is None else str(comment)
    raw = _strip_edges(raw)

    comment_id, text = _extract_comment_id(raw)

    # Extract bbox first so it doesn't pollute sections
    text, bbox = _extract_bbox(text)
    text = _strip_edges(text)

    # Strict fix payload (does not rely on segmentation)
    suggested_fix = extract_fix_payload(text)

    # Find expected and observed markers positions to segment cleanly
    m_exp = _find_earliest(text, EXP_MARKERS)
    m_obs = _find_earliest(text, OBS_MARKERS)
    m_fix = None
    # for segmentation boundaries, we only consider strict fix markers
    # (still okay to use them as boundaries)
    m_fix1 = RE_FIX_THIS.search(text)
    m_fix2 = RE_FIX_IT.search(text)
    if m_fix1 and m_fix2:
        m_fix = m_fix1 if m_fix1.start() < m_fix2.start() else m_fix2
    else:
        m_fix = m_fix1 or m_fix2

    markers = []
    if m_exp: markers.append(("expected", m_exp))
    if m_obs: markers.append(("observed", m_obs))
    if m_fix: markers.append(("fix", m_fix))
    markers.sort(key=lambda x: x[1].start())

    expected_standard = None
    observed_issue = None

    if markers:
        for i, (kind, m) in enumerate(markers):
            seg_start = m.end()
            seg_end = markers[i + 1][1].start() if i + 1 < len(markers) else len(text)
            seg = text[seg_start:seg_end]

            if kind == "expected":
                seg = _remove_expected_glue(seg)
                expected_standard = seg or None

            elif kind == "observed":
                # remove comma if exists + trim whitespace/newlines
                seg = _cleanup_segment(seg)
                observed_issue = seg or None

            # we do NOT populate suggested_fix from this segment;
            # we already did strict extraction above
    else:
        # If nothing recognized, keep everything as observed_issue (best fallback)
        observed_issue = _cleanup_segment(text) or None

    # If strict fix exists, and observed_issue accidentally includes "to fix this/it" tail, trim it.
    # (Not "hard stop": only removes the marker and anything after it from observed text.)
    if observed_issue:
        cut = None
        mm1 = RE_FIX_THIS.search(observed_issue)
        mm2 = RE_FIX_IT.search(observed_issue)
        if mm1 and mm2:
            cut = min(mm1.start(), mm2.start())
        elif mm1:
            cut = mm1.start()
        elif mm2:
            cut = mm2.start()
        if cut is not None:
            observed_issue = _cleanup_segment(observed_issue[:cut]) or None

    out = {
        "comment_id": comment_id,  # filled later if missing
        "expected_standard": expected_standard,
        "observed_issue": observed_issue,
        "suggested_fix": suggested_fix,
        "bounding_box": bbox,      # list[float] or None
        "raw_text": _norm_spaces(raw),
    }
    out["comment_type"] = infer_comment_type(out)
    return out


# ============================================================
# 4) Parse a cell (np.array / list / string) -> list[dict]
# ============================================================

def parse_comments_cell(cell: Any) -> List[Dict[str, Any]]:
    if cell is None or (isinstance(cell, float) and pd.isna(cell)):
        return []

    if isinstance(cell, np.ndarray):
        items = cell.tolist()
    elif isinstance(cell, (list, tuple)):
        items = list(cell)
    else:
        items = [cell]

    parsed = [parse_single_comment(x) for x in items]

    # Ensure a stable comment_id even if header missing
    for i, d in enumerate(parsed, start=1):
        if not d.get("comment_id"):
            d["comment_id"] = f"Comment {i}"
    return parsed


# ============================================================
# 5) Apply to your DataFrame + report IDs you couldn't extract
# ============================================================

def process_uicrit_comments(
    uicrit_df: pd.DataFrame,
    comments_col: str = "comments",
    screen_task_id_col: str = "screen_task_id",
    out_col: str = "parsed_comments",
) -> Tuple[pd.DataFrame, Dict[str, List[Any]]]:
    df = uicrit_df.copy()
    df[out_col] = df[comments_col].apply(parse_comments_cell)

    failures = {
        "missing_expected_standard": [],
        "missing_observed_issue": [],
        "missing_suggested_fix": [],
        "unclassified": [],
        "bbox_not_parsed_when_present": [],
    }

    for stid, plist in zip(df[screen_task_id_col].tolist(), df[out_col].tolist()):
        if not plist:
            failures["unclassified"].append(stid)
            continue

        miss_exp = miss_obs = miss_fix = any_unclassified = bbox_fail = False

        for d in plist:
            if d.get("expected_standard") is None:
                miss_exp = True
            if d.get("observed_issue") is None:
                miss_obs = True
            if d.get("suggested_fix") is None:
                miss_fix = True
            if d.get("comment_type") == "unclassified":
                any_unclassified = True

            raw_lower = (d.get("raw_text") or "").lower()
            if ("bounding box" in raw_lower or "boundingbox" in raw_lower) and d.get("bounding_box") is None:
                bbox_fail = True

        if miss_exp: failures["missing_expected_standard"].append(stid)
        if miss_obs: failures["missing_observed_issue"].append(stid)
        if miss_fix: failures["missing_suggested_fix"].append(stid)
        if any_unclassified: failures["unclassified"].append(stid)
        if bbox_fail: failures["bbox_not_parsed_when_present"].append(stid)

    # de-dup preserve order
    for k in failures:
        seen = set()
        failures[k] = [x for x in failures[k] if not (x in seen or seen.add(x))]

    return df, failures


# ============================================================
# 6) Usage
# ============================================================

uicrit_df, failures = process_uicrit_comments(uicrit_df, out_col="parsed_comments")
# print(failures.keys())
# print("Missing expected:", len(failures["missing_expected_standard"]))
# print("Missing observed:", len(failures["missing_observed_issue"]))
# print("Missing fix:", len(failures["missing_suggested_fix"]))
# print("Unclassified:", len(failures["unclassified"]))

# Example inspect:
# uicrit_df.loc[uicrit_df["screen_task_id"]=="68681", "parsed_comments"].iloc[0]
```

## Manually clean the parsed comments

```python
import re
from typing import List, Tuple, Dict, Any

# ------------------------------------------------------------
# 1) What to flag
#    - You said: flag if ANY field contains one or more of:
#      "expected standard" OR "current design" OR "to fix"
#    - Also keep the "\n" check if you want it
# ------------------------------------------------------------
FLAG_PAT = re.compile(
    # r"(expected\s+standard|current\s+design|to\s+fix|\n)",re.I
    r"(current\s+layout)",re.I
)

FIELDS = ("expected_standard", "observed_issue", "suggested_fix")

def field_has_flag(val: Any) -> bool:
    if val is None:
        return False  # IMPORTANT: don't auto-flag None (too noisy). We'll track None separately.
    return bool(FLAG_PAT.search(str(val)))

def field_is_none(val: Any) -> bool:
    return val is None

def should_flag_dict(d: Dict[str, Any], flag_none: bool = True) -> bool:
    """
    flag_none=True  -> also flag if any key is None (for cleanup)
    flag_none=False -> only flag if any key contains marker phrases/newlines
    """
    if any(field_has_flag(d.get(f)) for f in FIELDS):
        return True
    if flag_none and any(field_is_none(d.get(f)) for f in FIELDS):
        return True
    return False


# ------------------------------------------------------------
# 2) Collect flagged items (row index + comment index + reasons)
# ------------------------------------------------------------
def collect_flagged(uicrit_df, parsed_col="parsed_comments", flag_none=True):
    flagged: List[Tuple[int, int, Dict[str, Any]]] = []

    for row_i, row in uicrit_df.iterrows():
        plist = row.get(parsed_col, [])
        if plist is None or len(plist) == 0:
            continue

        for c_i, d in enumerate(plist):
            reasons = {"row_i": row_i, "c_i": c_i, "flags": {}, "nones": []}

            for f in FIELDS:
                v = d.get(f)
                if field_has_flag(v):
                    # store which terms triggered (rough)
                    triggers = sorted(set(re.findall(r"(expected\s+standard|current\s+design|to\s+fix|\n)", str(v), flags=re.I)))
                    reasons["flags"][f] = triggers
                if flag_none and v is None:
                    reasons["nones"].append(f)

            if reasons["flags"] or reasons["nones"]:
                flagged.append((row_i, c_i, reasons))

    return flagged


# ------------------------------------------------------------
# 3) Interactive editor:
#    - doesn't auto-advance after an edit
#    - lets you edit multiple keys in same dict before moving on
#    - supports 'raw' view, 'set none', and 'show current'
# ------------------------------------------------------------
def infer_comment_type(d: dict) -> str:
    has_exp = d.get("expected_standard") is not None
    has_obs = d.get("observed_issue") is not None
    has_fix = d.get("suggested_fix") is not None

    if has_exp and has_obs and has_fix:
        return "full"
    if has_exp and has_obs:
        return "expected_and_observed"
    if has_obs and has_fix:
        return "observed_and_fix"
    if has_exp and has_fix:
        return "expected_and_fix"
    if has_obs:
        return "observed_only"
    if has_fix:
        return "fix_only"
    if has_exp:
        return "expected_only"
    return "unclassified"


def inspect_and_edit(
    uicrit_df,
    flagged,
    parsed_col="parsed_comments",
    id_col="screen_task_id",
    start=0
):
    i = start
    n = len(flagged)

    help_txt = """
Commands:
  n                -> next
  p                -> previous
  q                -> quit
  raw              -> print raw_text
  show             -> show current extracted fields
  e / o / f        -> edit expected / observed / fix (prompts for new text)
  e+= / o+= / f+=  -> append text to that field
  none e|o|f       -> set field to None
  set e|o|f <text> -> set field to <text> (inline)
  help             -> show this help
Notes:
  - After edits, type 'show' to re-check before moving on.
  - comment_type is auto-updated after each edit.
""".strip()

    while 0 <= i < n:
        row_i, c_i, reasons = flagged[i]
        d = uicrit_df.at[row_i, parsed_col][c_i]

        print("\n" + "="*110)
        print(f"{i+1}/{n} | {id_col}={uicrit_df.at[row_i, id_col]} | comment_id={d.get('comment_id')} | comment_type={d.get('comment_type')}")
        print("-"*110)
        if reasons["flags"]:
            print("FLAGGED (field -> triggers):", reasons["flags"])
        if reasons["nones"]:
            print("NONE FIELDS:", reasons["nones"])
        print("Raw text:", d.get("raw_text"))
        print("-"*110)
        print("expected_standard :", d.get("expected_standard"))
        print("observed_issue    :", d.get("observed_issue"))
        print("suggested_fix     :", d.get("suggested_fix"))

        cmd = input("\ncmd (help for options): ").strip()

        if cmd == "q":
            break
        if cmd == "help":
            print("\n" + help_txt)
            continue
        if cmd == "n" or cmd == "":
            i += 1
            continue
        if cmd == "p":
            i = max(0, i - 1)
            continue
        if cmd == "raw":
            print("\nRAW_TEXT:\n" + (d.get("raw_text") or ""))
            continue
        if cmd == "show":
            # just reprint in next loop iteration
            continue

        # simple edit commands
        if cmd in {"e", "o", "f"}:
            field = {"e": "expected_standard", "o": "observed_issue", "f": "suggested_fix"}[cmd]
            new = input(f"New {field} (empty -> None):\n>>> ").rstrip("\n")
            d[field] = new.strip() if new.strip() else None
            d["comment_type"] = infer_comment_type(d)
            continue

        if cmd in {"e+=", "o+=", "f+="}:
            field = {"e+=": "expected_standard", "o+=": "observed_issue", "f+=": "suggested_fix"}[cmd]
            add = input(f"Append to {field}:\n>>> ").rstrip("\n")
            cur = d.get(field) or ""
            merged = (cur + (" " if cur and add else "") + add).strip()
            d[field] = merged if merged else None
            d["comment_type"] = infer_comment_type(d)
            continue

        # none e|o|f
        if cmd.startswith("none "):
            which = cmd.split(" ", 1)[1].strip().lower()
            field = {"e": "expected_standard", "o": "observed_issue", "f": "suggested_fix"}.get(which)
            if field:
                d[field] = None
                d["comment_type"] = infer_comment_type(d)
            else:
                print("Invalid: use none e|o|f")
            continue

        # set e|o|f <text>
        if cmd.startswith("set "):
            parts = cmd.split(" ", 2)
            if len(parts) < 3:
                print("Invalid: set e|o|f <text>")
                continue
            which = parts[1].strip().lower()
            text = parts[2]
            field = {"e": "expected_standard", "o": "observed_issue", "f": "suggested_fix"}.get(which)
            if not field:
                print("Invalid: set e|o|f <text>")
                continue
            d[field] = text.strip() if text.strip() else None
            d["comment_type"] = infer_comment_type(d)
            continue

        print("Unknown command. Type 'help'.")


# ------------------------------------------------------------
# 4) Usage
# ------------------------------------------------------------
# 4.1) Collect flagged items (choose whether to include None fields)
flagged = collect_flagged(uicrit_df, parsed_col="parsed_comments", flag_none=True)
print("Flagged items:", len(flagged))

# 4.2) Inspect + edit (press Enter or 'n' for next)
inspect_and_edit(uicrit_df, flagged, parsed_col="parsed_comments", id_col="screen_task_id")
```

```python
# Normalization and re-saving after edits
def normalize_comment_label(comment_id):
    if comment_id is None:
        return None, None

    comment_id = str(comment_id).strip()
    comment_source = "LLM" if re.match(r"^llm\b", comment_id, flags=re.I) else "Human"
    number_match = re.search(r"comment\s*(\d+)", comment_id, flags=re.I)

    if number_match:
        comment_label = f"Comment {number_match.group(1)}"
    else:
        comment_label = re.sub(r"^llm\s+", "", comment_id, flags=re.I).strip()

    return comment_label, comment_source

def normalize_bounding_box(bounding_box):
    if bounding_box is None:
        return None

    if isinstance(bounding_box, float) and pd.isna(bounding_box):
        return None

    if isinstance(bounding_box, dict):
        return {
            "x_min": float(bounding_box.get("x_min")) if bounding_box.get("x_min") is not None else None,
            "y_min": float(bounding_box.get("y_min")) if bounding_box.get("y_min") is not None else None,
            "x_max": float(bounding_box.get("x_max")) if bounding_box.get("x_max") is not None else None,
            "y_max": float(bounding_box.get("y_max")) if bounding_box.get("y_max") is not None else None,
        }

    if isinstance(bounding_box, np.ndarray):
        bounding_box = bounding_box.tolist()

    if isinstance(bounding_box, (list, tuple)) and len(bounding_box) >= 4:
        return {
            "x_min": float(bounding_box[0]),
            "y_min": float(bounding_box[1]),
            "x_max": float(bounding_box[2]),
            "y_max": float(bounding_box[3]),
        }

    return bounding_box

def has_content(value):
    if value is None:
        return False

    if isinstance(value, float) and pd.isna(value):
        return False

    return bool(str(value).strip())

def build_missing_parts(comment):
    comment_part_fields = ["expected_standard", "observed_issue", "suggested_fix"]
    return [field for field in comment_part_fields if not has_content(comment.get(field))]

def normalize_parsed_comments(parsed_comments):
    if parsed_comments is None:
        return parsed_comments

    if isinstance(parsed_comments, float) and pd.isna(parsed_comments):
        return parsed_comments

    if isinstance(parsed_comments, np.ndarray):
        parsed_comments = parsed_comments.tolist()

    normalized_comments = []

    for comment in parsed_comments:
        comment = dict(comment)
        comment_label, comment_source = normalize_comment_label(comment.pop("comment_id", None))
        comment.pop("comment_type", None)

        normalized_comment = {
            "comment_label": comment_label,
            "comment_source": comment_source,
            "missing_parts": build_missing_parts(comment),
            "bounding_box": normalize_bounding_box(comment.get("bounding_box")),
            "expected_standard": comment.get("expected_standard"),
            "observed_issue": comment.get("observed_issue"),
            "suggested_fix": comment.get("suggested_fix"),
            "raw_text": comment.get("raw_text"),
        }

        for key, value in comment.items():
            if key != "bounding_box" and key not in normalized_comment:
                normalized_comment[key] = value

        normalized_comments.append(normalized_comment)

    return normalized_comments

uicrit_df["parsed_comments"] = uicrit_df["parsed_comments"].apply(normalize_parsed_comments)

# Re-save the normalized structure so the parquet matches the updated schema.
output_path = Path("./cleaned_dataset") / "uicrit_notna_deduped.parquet"
uicrit_df.to_parquet(output_path, index=False)

uicrit_df["parsed_comments"].dropna().iloc[0][0]
```
