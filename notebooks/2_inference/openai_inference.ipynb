# OpenAI Inference

Runs the ratings or critiques stage with OpenAI-compatible models (OpenAI, Alibaba or DeepSeek, set by `PROVIDER`) on the shared schedules built in `0_prepare_schedules.ipynb`. Requests go through the provider's batch API; synchronous requests are an alternative for testing prompts or small samples.

# Imports And Notebook Root

```python
import sys
import json
from pathlib import Path
import pandas as pd
import os

ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "utils" / "paths.py").exists())
if str(ROOT) not in sys.path:
    sys.path.append(str(ROOT))

from utils.inference_utils import (
    CRITIQUES_SYSTEM_MESSAGE,
    EVALUATION_FIVE_ASPECTS,
    EVALUATION_MAIN_ASPECTS,
    GUIDELINES,
    RATING_SYSTEM_MESSAGE,
    run_llm_inference,
    append_response_jsonl,
    update_critiques_in_df,
    update_rating_in_df,
)
from utils.prompt_builder import CRITIQUE_PROMPTS, RATING_PROMPTS
from utils.data_setup import get_dataset_file_path, prepare_project_data
from utils.paths import RESULTS_DIR, SCHEDULES_DIR
from utils.models_setup import setup_client, query_openai_model
from utils.batch_manager import BatchConfig, BatchManager, OpenAIBatchAdapter
```

# Config And Constants

```python
TASK_SUBSET = "all_tasks"       # "all_tasks" | "1000_tasks"
PROMPTING_TYPE = "zero_shot"    # "zero_shot" | "few_shot"
STAGE = "critiques"             # "ratings" | "critiques"
NUM_TRIALS = 1
MAX_TOKENS = 30000
TEMPERATURE = 1
MODEL_SHORT = "gpt5"          # key in utils.models_setup.MODELS
PROMPT_VERSION = "p1"         # key in CRITIQUE_PROMPTS (critiques) or RATING_PROMPTS (ratings)
PROVIDER = "openai"          # "openai" | "alibaba" | "deepseek"

if STAGE == "ratings":
    SYSTEM_MESSAGE = RATING_SYSTEM_MESSAGE
    build_prompt_fn = RATING_PROMPTS[PROMPT_VERSION]["build_prompt_fn"]
    GUIDELINES = RATING_PROMPTS[PROMPT_VERSION]["guidelines"]
    update_results_df_fn = update_rating_in_df
elif STAGE == "critiques":
    SYSTEM_MESSAGE = CRITIQUES_SYSTEM_MESSAGE
    build_prompt_fn = CRITIQUE_PROMPTS[PROMPT_VERSION]["build_prompt_fn"]
    GUIDELINES = CRITIQUE_PROMPTS[PROMPT_VERSION]["guidelines"]
    update_results_df_fn = update_critiques_in_df
else:
    raise ValueError(f"Invalid stage: {STAGE}")

client, cfg = setup_client(
    provider=PROVIDER,
    model_short=MODEL_SHORT,
    system_message=SYSTEM_MESSAGE,
    temperature=TEMPERATURE,
    max_tokens=MAX_TOKENS,
)

cfg
```

# Data Loading

```python
uicrit_file, base64_screens_file, few_shot_samples_file = get_dataset_file_path(STAGE)

uicrit_df = pd.read_parquet(uicrit_file)
screenshots_df = pd.read_parquet(base64_screens_file)

if few_shot_samples_file:
    few_shot_samples_df = pd.read_parquet(few_shot_samples_file)
else:
    few_shot_samples_df = None

print("UICrit rows:", len(uicrit_df))
print("Screens rows:", len(screenshots_df))
```

# Select Schedule For The Current Run

```python
schedule_dir = SCHEDULES_DIR

rating_schedule_file_1000_tasks = schedule_dir / "rating_schedule-1000_tasks.parquet"
rating_schedule_file_all_tasks = schedule_dir / "rating_schedule-all_tasks.parquet"
critiques_schedule_file_1000_tasks = schedule_dir / "critiques_schedule-1000_tasks.parquet"
critiques_schedule_file_all_tasks = schedule_dir / "critiques_schedule-all_tasks.parquet"
critiques_schedule_file_all_tasks_excluding_1000 = schedule_dir / "critiques_schedule-all_tasks_excluding_1000.parquet"

if TASK_SUBSET == "1000_tasks":
    if STAGE == "ratings":
        schedule_file = rating_schedule_file_1000_tasks
    else:
        schedule_file = critiques_schedule_file_1000_tasks
else:
    if STAGE == "ratings":
        schedule_file = rating_schedule_file_all_tasks
    else:
        schedule_file = critiques_schedule_file_all_tasks

print("Using schedule file:", schedule_file)
```

# Load Responses DataFrame

```python
responses_df, results_paths = prepare_project_data(
    model_short=cfg["model_short"],
    num_trials=NUM_TRIALS,
    task_subset=TASK_SUBSET,
    shots=PROMPTING_TYPE,
    stage=STAGE,
)

text_responses_jsonl_file = results_paths["results_jsonl"]
model_results_file = results_paths["results_parquet"]

print("Responses shape:", responses_df.shape)
responses_df.head(2)
```

# Batch Requests

Send the schedule through the provider's batch API.

## Batch Setup

```python
batch_main_dir = RESULTS_DIR / STAGE / "inference_batches" / f"Batch_Files-{TASK_SUBSET}-{PROMPTING_TYPE}-{cfg['model_short']}"
batch_main_dir.mkdir(parents=True, exist_ok=True)

batch_config = BatchConfig(
    provider=cfg["provider"],
    model=cfg["model_version"],
    batch_main_dir=batch_main_dir,
    selected_tasks=TASK_SUBSET,
    temperature=TEMPERATURE,
    max_tokens=MAX_TOKENS,
    system_message=SYSTEM_MESSAGE,
    image_format="jpeg",
)

batch_process = BatchManager(OpenAIBatchAdapter(client, batch_config))

batch_main_dir
```

## Prepare Request Batch Files

```python
if any(batch_process.requests_dir.glob(f"requests-{TASK_SUBSET}-batch_*.jsonl")):
    print("Request files already exist (delete them to rebuild):", batch_process.requests_dir)
else:
    batch_process.generate_request_batches_jsonl(
        schedule_file=schedule_file,
        responses_df=responses_df,
        screens_df=screenshots_df,
        guidelines=GUIDELINES,
        prompting_type=PROMPTING_TYPE,
        samples_df=few_shot_samples_df,
        build_prompt_fn=build_prompt_fn,
        max_mb_per_file=180,
    )

batch_process.preview_request(index=0)
```

## Submit And Retrieve Batches

Uploads, submits, polls and downloads each request batch file in turn. Batch files whose responses are
already downloaded are skipped, and a batch still running at the provider is resumed from the jobs manifest.

```python
start_from_batch = 1
resume_batch = None

batch_process.process_batches_sequentially(
    start_from_batch=start_from_batch,
    sleep_minutes=2,
    resume_batch_id=resume_batch,
)
```

## Extract Results From Batch Responses

```python
summary = batch_process.extract_results_from_responses(
    responses_df,
    update_results_df_fn,
)

print(
    "Updated:", summary["updated"],
    "| Skipped:", len(summary["skipped"]),
    "| Errors:", len(summary["errors"]),
)

responses_df.head(2)
```

```python
errors = batch_process.inspect_response_errors()
if errors:
    errors[:5]
```

## Retry Failed Requests

```python
failed_custom_ids = [error["custom_id"] for error in errors]

if not failed_custom_ids:
    print("No failed requests to retry.")
else:
    # Each retry round gets its own files, so earlier retry responses are kept; retry files are read last.
    retry_round = len(list(batch_process.responses_dir.glob("responses-retry_failed*.jsonl"))) + 1
    retry_base_name = f"responses-retry_failed_{retry_round}"
    retry_requests_file = batch_process.requests_dir / f"requests-retry_failed_{retry_round}.jsonl"

    batch_process.build_retry_requests_from_failed_requests(
        failed_custom_ids,
        source_requests_file=batch_process.requests_dir,
        output_file=retry_requests_file,
    )
    retry_input_file_id = batch_process.upload_file(retry_requests_file)
    batch_process.create_batch(input_file_id=retry_input_file_id)
    batch_process._poll_then_retrieve(sleep_minutes=2, base_name=retry_base_name)

    retry_extract_summary = batch_process.extract_results_from_responses(
        responses_df,
        update_results_df_fn,
        pattern=f"{retry_base_name}*.jsonl",
    )
    print("Updated:", retry_extract_summary["updated"], "| Skipped:", len(retry_extract_summary["skipped"]))
```

# Alternative: Synchronous Requests

Use this section instead of batching when you want to test prompts or run a very small sample interactively.

Set `RUN_ONE_BY_ONE = True` to run it; it is off so that Run All never sends these requests.

```python
RUN_ONE_BY_ONE = False  # True sends paid one-by-one requests for rows still without a response

if RUN_ONE_BY_ONE:
    query_args = dict(
        client=client,
        model_version=cfg["model_version"],
        max_tokens=MAX_TOKENS,
        temperature=TEMPERATURE,
        system=SYSTEM_MESSAGE,
        image_format="jpeg",
    )

    run_llm_inference(
        responses_df=responses_df,
        screenshots_df=screenshots_df,
        few_shot_samples_df=few_shot_samples_df,
        evaluation_aspects=EVALUATION_MAIN_ASPECTS,
        build_prompt_fn=build_prompt_fn,
        update_results_df_fn=update_results_df_fn,
        query_fn=query_openai_model,
        query_args=query_args,
        save_jsonl_fn=append_response_jsonl,
        output_jsonl=text_responses_jsonl_file,
        guidelines=GUIDELINES,
        prompting_type=PROMPTING_TYPE,
        requests_per_minute=cfg["rpm"],
        stage=STAGE,
        print_output=False,
    )
```
